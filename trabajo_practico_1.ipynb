```python
# ejercicio 4

#esta clase se encarga de realizar los calculos del ejercicio
class calcular_edades:
  def calcular_edalber(edjuan):
    return (2 * edjuan) / 3    #formula para calcular la edad de alberto


  def calcular_edana(edjuan):
    return (4 * edjuan) / 3    #formula para calcular la edad de ana


  def calcular_edmama(edjuan , edalber , edana):
    return edjuan + edalber + edana   #formula para calcular la edad de la mama

edjuan = int(input("ingrese la edad de juan:"))
edalber = calcular_edades.calcular_edalber(edjuan)
edana = calcular_edades.calcular_edana(edjuan)
edmama = calcular_edades.calcular_edmama(edjuan , edalber , edana)

# resultado
print(f"las edades son : Alberto = {edalber} , Juan = {edjuan} , Ana = {edana} , Mamá = {edmama}")
```

```python
# ejercicio 5
class calculo:
  def __init__(self, x_inicial , y_inicial):
    self.suma = 0
    self.x = x_inicial
    self.y = y_inicial

  def ejecutar(self):
    self.suma += self.x  #suma = suma + x
    self.x += self.y ** 2  #x = x + y ** 2
    self.suma += self.x / self.y  #suma = x /y
    return self.suma

#objeto

proceso = calculo(20 , 40)  # valores que toman x e y respectivamente
print(f"El valor inicial de x es: {proceso.x}")
print(f"El valor inicial de y es: {proceso.y}")
print(f"El valor de la suma es: {proceso.ejecutar()}")
```

```python
# ejercicio 12
class nomina:
  def cal_salario_bruto(horas_trabajadas , valor_hora):
    return horas_trabajadas * valor_hora

  def cal_retencion(salario_bruto , porcentaje_retencion):
    return salario_bruto * porcentaje_retencion

  def cal_salario_neto(salario_bruto , valor_retencion):
    return salario_bruto - valor_retencion

# definicion de las variables
horas_trabajadas = 48
valor_hora = 5000
retencion = 12.5
porcentaje_retencion = retencion / 100

salario_bruto = nomina.cal_salario_bruto( horas_trabajadas , valor_hora )
valor_retencion = nomina.cal_retencion( salario_bruto , porcentaje_retencion )
salario_neto = nomina.cal_salario_neto( salario_bruto , valor_retencion )

print(f"el salario bruto es: ${salario_bruto}")
print(f"el valor de retencion en la fuente es: ${valor_retencion}")
print(f"el salario neto es: ${salario_neto}")
```

```python
# ejercicio 14
class cal_potencias:
  def cal_cuadrado(numero):
    return numero ** 2

  def cal_cubo(numero):
    return numero ** 3

numero = float(input("ingrese un número: "))

cuadrado = cal_potencias.cal_cuadrado(numero)
cubo = cal_potencias.cal_cubo(numero)

print(f"el cuadrado de {numero} es: {cuadrado}")
print(f"el cubo de {numero} es: {cubo}")
```

```python
#ejercicio 17

# importamos primero la libreria math para poder hacer los calculos
import math

class cal_circulo:
  def cal_area(radio):        # area del circulo = pi * radio ** 2
    return math.pi * radio ** 2

  def cal_perimetro(radio):   # longitud = 2 * pi * radio
    return 2 * math.pi * radio

radio = float(input(" ingrese el radio del circulo: "))

area = cal_circulo.cal_area(radio)
longitud = cal_circulo.cal_perimetro(radio)

print(f"el area del circulo es: {area:.2f}")
print(f"la longitud del circulo es: {longitud:.2f}")
```
